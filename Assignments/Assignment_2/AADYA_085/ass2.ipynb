```python
#Find sum of first 10 natural numbers.
sum=0
for i in range(1,11):
  sum+=i
print(sum)
```

```python
#Find factorial of a number.
n=int(input("Enter a number for calculating factorial : "))
fact=1
for i in range (1,n+1):
  fact*=i
print(fact)
```
