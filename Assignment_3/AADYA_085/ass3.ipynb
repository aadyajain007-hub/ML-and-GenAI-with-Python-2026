```python
#Create a function to print first 10 natural numbers.
def natural(n):
  for i in range(n+1):
    print(i)
natural(10)
```

```python
#Create a function to calculate sum of first N natural numbers.
def sum_nat(n):
  sum=0
  for i in range(n+1):
    sum+=i
  print(sum)
sum_nat(25)
```

```python
#Create a function to reverse a number.
def rev(n):
  return int(str(n)[::-1])
print(rev(8501042025))
```

```python
#Create a function to count digits in a number.
def count(n):
  return len((str(n)))
print(count(11234568790))
```

```python
#Create a function to check palindrome number.
def pal(n):
  return str(n)==str(n)[::-1]
print(pal(112211))
```

```python
#Create a function to generate Fibonacci series.
def fib(n):
  a=0
  b=1
  for i in range(n+1):
    print(a)
    c=a+b
    a=b
    b=c
fib(10)
```

```python
#Calculator Using Functions that contains the following features;
#	-	User selects operation
#	-	Program performs calculation
#	-	Display result
def add(a,b):
	return a+b
def sub(a,b):
	return a-b
def mul(a,b):
	return a*b
def div(a,b):
	if b==0:
		print("cannot be divided")
	return a/b
print("1.add")
print("2.sub")
print("3.multiply")
print("4.divide")
choice=int(input("Choose one of the following"))
a=int(input("Enter first number"))
b=int(input("Enter second number"))
if choice==1:
	print(add(a,b))
elif choice==2:
	print(sub(a,b))
elif choice==3:
	print(mul(a,b))
elif choice==4:
	print(div(a,b))
else:
	print("invalid choice")
```

```python
#Create a student class with name and marks
class student:
    def __init__(self, name, marks):
        self.name = name
        self.marks = marks

    def display(self):
        print("Name:", self.name)
        print("Marks:", self.marks)

name = input("Enter student's name: ")
marks = int(input("Enter student's marks: "))

s = student(name, marks)

s.display()
```
